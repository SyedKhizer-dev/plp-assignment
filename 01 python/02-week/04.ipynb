```python
import pandas as pd
```

```python

student_df = pd.read_csv("data/student.csv")
scores_df = pd.read_csv("data/score.csv")
print(student_df)
scores_df
```

```python
merge_df = pd.merge(student_df, scores_df, on="student_id", how="inner")
merge_df
```

```python
result = merge_df.groupby("subject")["score"].mean()
result1 = merge_df.groupby("name")["score"].mean()
print("Group by subject:\n", result)
print("Group by student:\n", result1)
```

```python
def grade_generator(score):
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 55:
        return "C"
    else:
        return "F"


merge_df["grade"] = merge_df["score"].apply(grade_generator)
merge_df
```

```python
import numpy as np
```

```python
mean_value=np.mean(merge_df["score"])
std_value=np.std(merge_df["score"])
per_value=np.percentile(merge_df["score"],75)
print(mean_value)
print(std_value)
print(per_value)
```
